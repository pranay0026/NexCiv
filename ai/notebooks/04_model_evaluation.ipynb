# 🏛️ NexCiv AI: 04. Model Evaluation & Disaggregated Language Benchmark
### Evaluating Pure Telugu, English, and Romanized Telugu Separately
---
### 🔬 Critical Experiment:
> **Train on all three styles and test separately on:**
> 1. **Pure Telugu (`te`)**
> 2. **Pure English (`en`)**
> 3. **Romanized Telugu (`te_latn`)**
> This demonstrates whether NexCiv actually understands all three writing styles rather than relying on surface memorization.

```python
import os
import pandas as pd
import joblib
from sklearn.metrics import classification_report, accuracy_score, f1_score

test_p = '../datasets/test.csv' if os.path.exists('../datasets/test.csv') else 'ai/datasets/test.csv'
test_df = pd.read_csv(test_p)

model_p = '../../models/complaint_classifier/complaint_classifier.pkl'
if not os.path.exists(model_p):
    model_p = 'models/complaint_classifier/complaint_classifier.pkl'

bundle = joblib.load(model_p)
dept_pipe = bundle['wing_pipeline']
type_pipe = bundle.get('complaint_type_pipeline')

print("=" * 65)
print("DISAGGREGATED PERFORMANCE ACROSS LANGUAGE STYLES")
print("=" * 65)

for lang in ['te', 'en', 'te_latn']:
    sub = test_df[test_df['language_style'] == lang]
    d_preds = dept_pipe.predict(sub['complaint_text'])
    d_acc = accuracy_score(sub['department'], d_preds)
    d_f1 = f1_score(sub['department'], d_preds, average='macro', zero_division=0)
    print(f"Language Style [{lang:7}] | Department Acc: {d_acc*100:5.1f}% | Macro F1: {d_f1:.4f} (Samples: {len(sub)})")
```
